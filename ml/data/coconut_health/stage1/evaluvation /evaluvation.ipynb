```python

# CocoCare: Weather data quality evaluation

import pandas as pd
import numpy as np
from IPython.display import display

# Parse dates
weather_df["date"] = pd.to_datetime(
    weather_df["date"], errors="coerce"
)

# 1. Basic checks
print("DATASET SHAPE:", weather_df.shape)
print("DATE RANGE:", weather_df["date"].min(),
      "to", weather_df["date"].max())
print("DUPLICATE ROWS:", weather_df.duplicated().sum())
print("DUPLICATE DATES:", weather_df["date"].duplicated().sum())

# 2. Missing values
missing = pd.DataFrame({
    "missing_count": weather_df.isna().sum(),
    "missing_percent": (
        weather_df.isna().mean() * 100
    ).round(2)
})
print("\nMISSING VALUE REPORT")
display(missing)

# 3. Check calendar coverage
expected_dates = pd.date_range(
    "2023-01-01", "2023-12-31", freq="D"
)
observed_dates = pd.DatetimeIndex(
    weather_df["date"].dropna().dt.normalize().unique()
)

missing_dates = expected_dates.difference(observed_dates)
print("\nEXPECTED DAYS:", len(expected_dates))
print("UNIQUE OBSERVED DAYS:", len(observed_dates))
print("MISSING CALENDAR DAYS:", len(missing_dates))
print("FIRST MISSING DATES:", missing_dates[:10].tolist())

# 4. Physical-range screening
ranges = {
    "rainfall_mm": (0, 500),
    "tmax_c": (-10, 60),
    "tmin_c": (-15, 50),
    "solar_radiation_mj_m2_day": (0, 45),
    "relative_humidity_pct": (0, 100),
    "wind_speed_m_s": (0, 75)
}

print("\nRANGE CHECKS")
for col, (low, high) in ranges.items():
    values = pd.to_numeric(weather_df[col], errors="coerce")
    count = ((values < low) | (values > high)).sum()
    print(f"{col}: {count} values outside screening range")

# 5. Temperature consistency
bad_temp = weather_df["tmax_c"] < weather_df["tmin_c"]
print("\nDAYS WHERE TMAX < TMIN:", int(bad_temp.sum()))

# 6. Numeric summary
print("\nNUMERIC SUMMARY")
display(
    weather_df[list(ranges)].describe().T.round(3)
)
```

```python

# CocoCare Stage 1: Distribution checks

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

# hybrid_df was loaded in your previous cell
df_eval = hybrid_df.copy()
df_eval["date"] = pd.to_datetime(df_eval["date"], errors="coerce")

print("Rows:", len(df_eval))
print("Columns:", len(df_eval.columns))
print("Unique dates:", df_eval["date"].nunique())
print("Duplicate dates:", df_eval["date"].duplicated().sum())
print("Date range:", df_eval["date"].min(), "to", df_eval["date"].max())

print("\nScenario distribution:")
display(df_eval["scenario"].value_counts(dropna=False).to_frame("count"))

print("\nWater stress label distribution:")
display(
    df_eval["water_stress_rule_derived"]
    .value_counts(dropna=False)
    .to_frame("count")
)

print("\nRecommended action distribution:")
display(
    df_eval["recommended_action_simulated"]
    .value_counts(dropna=False)
    .to_frame("count")
)

numeric_cols = [
    "et0_mm_day",
    "etc_potential_mm_day_modelled",
    "effective_rainfall_mm_modelled",
    "runoff_mm_modelled",
    "irrigation_mm_simulated",
    "actual_et_mm_day_modelled",
    "drainage_mm_modelled",
    "soil_water_storage_mm_modelled",
    "depletion_fraction_modelled",
    "soil_moisture_vwc_pct_simulated",
]

display(df_eval[numeric_cols].describe().T.round(3))
```

```python

# CocoCare Stage 1: Basic physical-consistency checks

checks = []

def add_check(name, count, explanation):
    checks.append({
        "check": name,
        "flagged_rows": int(count),
        "interpretation": explanation
    })

# Non-negative quantities
nonnegative_cols = [
    "rainfall_mm",
    "solar_radiation_mj_m2_day",
    "wind_speed_m_s",
    "et0_mm_day",
    "etc_potential_mm_day_modelled",
    "effective_rainfall_mm_modelled",
    "runoff_mm_modelled",
    "irrigation_mm_simulated",
    "actual_et_mm_day_modelled",
    "drainage_mm_modelled",
    "soil_water_storage_mm_modelled",
]

for col in nonnegative_cols:
    values = pd.to_numeric(df_eval[col], errors="coerce")
    add_check(
        f"{col}: negative values",
        (values < -1e-8).sum(),
        "Investigate negative values; do not silently clip them."
    )

# Temperature consistency
add_check(
    "Maximum temperature below minimum",
    (df_eval["tmax_c"] < df_eval["tmin_c"]).sum(),
    "Maximum temperature should not be below minimum temperature."
)

# Relative humidity bounds
rh = df_eval["relative_humidity_pct"]
add_check(
    "Relative humidity outside 0–100%",
    ((rh < 0) | (rh > 100)).sum(),
    "Check source units and missing-value handling."
)

# Depletion fraction is normally bounded by the model's definition.
depletion = df_eval["depletion_fraction_modelled"]
add_check(
    "Depletion fraction outside 0–1",
    ((depletion < -1e-6) | (depletion > 1 + 1e-6)).sum(),
    "Check the depletion formula and clipping rules."
)

# Simulated volumetric water content percentage
vwc = df_eval["soil_moisture_vwc_pct_simulated"]
add_check(
    "Simulated VWC outside 0–100%",
    ((vwc < 0) | (vwc > 100)).sum(),
    "Check the conversion from storage to volumetric water content."
)

consistency_report = pd.DataFrame(checks)
display(consistency_report)
```

```python

# CocoCare: daily water-balance consistency check

storage_col = "soil_water_storage_mm_modelled"

# The current configuration appears to use 58.5 mm as the
# maximum available storage. Confirm this against metadata/code.
TAW_ASSUMED = 58.5

balance = df_eval.sort_values("date").copy()

previous_storage = balance[storage_col].shift(1)

balance["storage_before_clipping_mm"] = (
    previous_storage
    + balance["effective_rainfall_mm_modelled"]
    + balance["irrigation_mm_simulated"]
    - balance["actual_et_mm_day_modelled"]
    - balance["runoff_mm_modelled"]
    - balance["drainage_mm_modelled"]
)

balance["storage_predicted_mm"] = (
    balance["storage_before_clipping_mm"]
    .clip(lower=0, upper=TAW_ASSUMED)
)

balance["storage_error_mm"] = (
    balance[storage_col] - balance["storage_predicted_mm"]
)

# First row is excluded because its previous day's storage
# is not available in this dataset.
checked = balance.iloc[1:].copy()

print("Days checked:", len(checked))
print(
    "Mean absolute storage error (mm):",
    checked["storage_error_mm"].abs().mean()
)
print(
    "Maximum absolute storage error (mm):",
    checked["storage_error_mm"].abs().max()
)
print(
    "Days with error > 0.01 mm:",
    (checked["storage_error_mm"].abs() > 0.01).sum()
)

display(
    checked.loc[
        checked["storage_error_mm"].abs() > 0.01,
        [
            "date",
            "soil_water_storage_mm_modelled",
            "storage_predicted_mm",
            "storage_error_mm"
        ]
    ].head(20)
)
```

```python

# Check the implied Kc from the stored ET0 and ETc values

valid_et0 = df_eval["et0_mm_day"] > 1e-8

implied_kc = (
    df_eval.loc[valid_et0, "etc_potential_mm_day_modelled"]
    / df_eval.loc[valid_et0, "et0_mm_day"]
)

print(implied_kc.describe())

print(
    "\nMedian implied Kc:",
    round(implied_kc.median(), 4)
)
```

```python

# CocoCare: investigate the largest water-balance discrepancies

cols = [
    "date",
    "rainfall_mm",
    "effective_rainfall_mm_modelled",
    "runoff_mm_modelled",
    "irrigation_mm_simulated",
    "etc_potential_mm_day_modelled",
    "actual_et_mm_day_modelled",
    "drainage_mm_modelled",
    "soil_water_storage_mm_modelled",
    "storage_predicted_mm",
    "storage_error_mm",
]

diagnostic = balance[cols].copy()

# Exclude the first day because previous-day storage is unknown
diagnostic = diagnostic.iloc[1:].copy()

diagnostic["abs_error_mm"] = diagnostic["storage_error_mm"].abs()

display(
    diagnostic.sort_values(
        "abs_error_mm", ascending=False
    ).head(20)
)
```

```python

# Diagnostic only: test whether runoff is already excluded
# from effective rainfall. Do not change the generator yet.

balance["storage_predicted_no_double_runoff_mm"] = (
    previous_storage
    + balance["effective_rainfall_mm_modelled"]
    + balance["irrigation_mm_simulated"]
    - balance["actual_et_mm_day_modelled"]
    - balance["drainage_mm_modelled"]
).clip(lower=0, upper=TAW_ASSUMED)

balance["error_no_double_runoff_mm"] = (
    balance["soil_water_storage_mm_modelled"]
    - balance["storage_predicted_no_double_runoff_mm"]
)

diagnostic_rows = balance.iloc[1:]

print(
    "Mean absolute error without second runoff subtraction:",
    diagnostic_rows["error_no_double_runoff_mm"].abs().mean()
)

print(
    "Maximum absolute error without second runoff subtraction:",
    diagnostic_rows["error_no_double_runoff_mm"].abs().max()
)

print(
    "Days with error > 0.01 mm:",
    (
        diagnostic_rows["error_no_double_runoff_mm"].abs() > 0.01
    ).sum()
)

display(
    diagnostic_rows[
        [
            "date",
            "rainfall_mm",
            "effective_rainfall_mm_modelled",
            "runoff_mm_modelled",
            "soil_water_storage_mm_modelled",
            "storage_predicted_no_double_runoff_mm",
            "error_no_double_runoff_mm"
        ]
    ].iloc[:10]
)
```

```python

# Final water-balance arithmetic validation

balance = df_eval.sort_values("date").copy()

TAW_ASSUMED = 58.5  # Confirm against generator configuration

previous_storage = (
    balance["soil_water_storage_mm_modelled"].shift(1)
)

predicted_storage = (
    previous_storage
    + balance["effective_rainfall_mm_modelled"]
    + balance["irrigation_mm_simulated"]
    - balance["actual_et_mm_day_modelled"]
    - balance["drainage_mm_modelled"]
).clip(lower=0, upper=TAW_ASSUMED)

observed_storage = balance["soil_water_storage_mm_modelled"]

error = observed_storage - predicted_storage
valid_error = error.iloc[1:].dropna()

print("Days evaluated:", len(valid_error))
print("Mean absolute error (mm):", valid_error.abs().mean())
print("Maximum absolute error (mm):", valid_error.abs().max())
print("Days with error > 0.01 mm:",
      (valid_error.abs() > 0.01).sum())

assert valid_error.abs().max() < 1e-8, (
    "Water-balance mismatch: inspect generator logic."
)

print("\nPASS: Stored water balance matches the reconstructed equation.")
```
